# Training Commands

Run the setup cell first so relative paths resolve from the repository root. The `%%bash` cells below are the runnable training commands.


````python
from pathlib import Path
import os

root = Path.cwd().resolve()
while root != root.parent and not (root / "src/train/commands.md").exists():
    root = root.parent

if not (root / "src/train/commands.md").exists():
    raise FileNotFoundError("Could not find repository root containing src/train/commands.md")

os.chdir(root)
print(f"Working directory: {root}")
````

## Before Training

- Single-label runs require extracted feature manifests in `data/processed`.
- Multi-label runs require mixed manifests in `data/processed`.

````python
%%bash
python -m src.train.run_train \
  --config src/configs/training/irmas/single_label_mel.yaml \
  --output_dir src/models/saved_weights/my_run
````

## Fine-Tune From a Previous Checkpoint

Set `model.pretrained_weights` inside the target YAML, then run it normally:

```yaml
model:
  backbone: "cnn"
  dropout: 0.3
  pretrained_weights: "src/models/saved_weights/irmas_single_label_mel/best_val.pt"
```


````python
%%bash
python -m src.train.run_train \
  --config src/configs/training/chinese_instruments/single_label_mel.yaml
````

## Pattern for New Training YAMLs

If you add another preset under `src/configs/training/...`, call it the same way:

```bash
python -m src.train.run_train \
  --config src/configs/training/<dataset>/<your_config>.yaml
```
